# Tensor decomposition of spherical harmonics.

Did you know that a degree-$\ell$ spherical harmonic is the same thing as a traceless symmetric rank-$\ell$ tensor? You don't understand what it means? Me neither, this is why I break it here.

## What is a tensor? 

A tensor is group built over a field ($\mathbb{R}$, $\mathbb{C}$...) that comes after a matrix : 
- a *scalar* is a single value of the field (rank = 0)
- a *vector* is a 1-set of values (rank = 1). Values can be noted $T_i$
- a *matrix* is a 2-set of values (rank = 2). values can be noted $T_{ij}$.
- a *tensor* is everything after. Values can be noted $T_{ijk}$. 

A tensor of rank 3 can then have a shape $3\times3\times3$. What makes it a tensor is the tensor product : for example, under a rotation of space, each index transforms like a vector. A rank-2 tensor rotates as $T \mapsto R T R^T$ (both indices rotate), and a rank-$k$ tensor rotates each of its $k$ slots the same way. 

**Einstein summation convention**. In an equation, if an indice is not present in both sides, it means it is summed. For example :  

- matrix x vector = $(Av)_i = A_{ij}v_j$ -> (j summed)
- matrix x matrix = $(AB)_{ik} = A_{ij}B_{jk}$ -> (j summed)

**Kronecker delta.** A special operator is the Kronecked delta, defined by 

$$
\delta_{ij} = 
\begin{cases}
 & 0 \text{ if } i\neq j\\
 & 1 \text{ if } i=j\\
\end{cases}
$$

$\delta_{ij}$ can be used for 
- *dot products* : $\delta_{ij}v_j = v_i$
- *rank values* : $\delta_{ii}$

It is rotation-invariant : the identity matrix loks the same in every rotated frame, such that: $R \delta R^T = \delta$. 

**Contraction and trace.** A *contraction* is a generalisation of the *trace* operation for matrices that lowers the rank by 2. Trace can be indeed written : 
$$
T_{ii} = \delta_{ij}T_{ij}
$$
as $\delta$ is rotation invariant, the trace of a rotated tensor is also rotation invariant. 

For higher ranks we can contract any pair of indices : T_{iik} contracts the first two, and leaves a rank-1 tensor. 

**Symmetric tensors** A tensor is symmetric if 
- $T_{ij} = T_{ji}$ (rank-2)
- $T_{ijk} = T_{jik} = T_{ikj} = \dots$ (unchanged under any indice permutation)



## Making tensors out of vectors (outer product)

The **outer product** is the rank-2 tensor
$$
T_{ij} = u v^T = u_i v_j = \begin{pmatrix}
u_1v_1 & u_1 v_2 & u_1 v_3 \\
u_2v_1 & u_2 v_2 & u_2 v_3 \\
u_3v_1 & u_3 v_2 & u_3 v_3 \\
\end{pmatrix}
$$

and over : $T_{ijk} = u_iv_jw_k$, and so on. Here we are interested by **a direction multiplied by itself** : for a diretion $\hat n$ : 
- $n_in_j$ (rank 2)
- $n_in_jn_K$ (rank 3)
- $\dots$
- $n_in_j\dots n_\ell$ (rank $\ell$)

these are **the build blocks of spherical harmonics**. They are automatically symmetric through commutative multiplication in the field. 

A symmetric is traceless if all its traces vanish. For rank-2, this implies $T_{ii} = 0$. Any symmetric matrix splits **uniquely** into a traceless part plus a trace part : 
$$T_{ij} = \underbrace{(T_{ij} - \frac{1}{3}\delta_{ij} T_{kk})}_{\text{traceless}} + \underbrace{\frac{1}{3}\delta_{ij}T_{kk}}_{\text{trace part}}$$
(for indices for 1 to 3, as we're interested here). As $6 = 5 + 1$, the traceless symmetric part has 5 components and the trace part is one number ($T_{kk}$) times the fixed tensor $\delta$. 

For a general rank $\ell$, tracelss means *every* pairwaise contraction is zero. The split is then 
$$
(\text{symmetric rank } \ell) = (\text{traceless rank }\ell) \oplus (\text{symmetric rank } \ell - 2) \oplus \delta^2 \cdot (\text{rank } \ell - 4) \oplus \dots
$$

substracting the trace part leaves exactly 
$$
(\ell+1)(\ell+2)/2 - (\ell-1)\ell/2 = 2\ell+1 
$$
components. We cann see that rotations not only preserve symmetry (swapping indices commtes with rotating them), but also preserves tracelessness (as trace is rotation invariant). Hence, the traceless part has the dimensions of $\mathcal{H}^l$, the number of spherical harmonics of degree $\ell$. Is that a coincidence? 



## Link to spherical harmonics

Because rotations preserve symmetry and tracelessness, the set of traceless symmetric rank-$\ell$ tensors is a subspace that rotations map into itself : an **invariant subspace** of dimension $2\ell + 1$. It is *irreductible* : you cannot split it into smaller rotation-invariant pieces. We can also say that this is the "spin-$\ell$ irreducible representation of the rotation group", that is exactly what a degree $\ell$ of spherical harmonics is. 

**Tensors $\leftrightarrow$ polynomials**. A symmetric rank-$\ell$ tensor $T$ and a homogenous polynomial of degree $\ell$ are the same data bia contraction with $\mathbf{x} = (x, y, z) 
$$
p(x) = T_{i_1\dots i_\ell} x_{i_1} \dots x_{i_\ell}
$$

For example, in rank 2, $p(x) = T_{ij}x_{i}x_{j} = x^T T x$ : a quadratic form. On the unit sphere $\mathbf{x} \rightarrow \hat n$, this is a function of direction : a canditate spherical harmonic. 

**Laplacian is the trace**. The Laplacian is defined is $\nabla^2 = \delta_i \delta_j$ (also a contraction of two derivatives). Applying to $p$ : 

- rank 2: $\nabla^2(T_{ij} x_i x_j) = 2 T_{ii}$, hence $2\mathrm{Tr}(T)$
- rank 3: $\nabla^2{T_{ijk} x_i x_j x_k} = 2 T_{iik}x_k$, hence $6\cdot \mathrm{Tr}(T)\cdot \mathbf{x}$

Hence, differentiating twice and summing $\leftrightarrow$ contracting two indices $\leftrightarrow$ taking a trace. Therefore, **$p$ is harmonic $\leftrightarrow$ $T$ is traceless.** Hence, a harmonic polynomial is precisely a traceless symemtric tensor. Restricted to the sphere, harmonic polynomials of degree $\ell$ are exactly the spherical harmonics $Y_\ell^m$, and there are $2\ell+1$ of them.
