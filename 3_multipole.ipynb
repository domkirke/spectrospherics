# Multipoles

Same but not same, multipoles are an alternative to spherical harmonics. They also have another correspondancy with another paramterization of the spherical expansion : *Maxwell axes*.

## Maxwell axes

Represent a degree $\ell$-harmonic by $\ell$ axes $\hat{\mathbf{u}}, \dots, \mathbf{\hat{u}}_\ell$ plus a magnitude $r$.

**Raw (reducible) version.** The plain product of directional responses is then: 
$$g_{raw}(\hat{n}) = r \cdot \prod_{k=1}^{\ell}(\hat{n} \cdot \hat{u}_k)$$
That is a homogeneous degree-$\ell$ polynomial, so on the sphere it is a sum of harmonics of degrees $\ell, \ell-2, \ell-4, \dots$ (same parity) ; not pure degree $\ell$. 

**Pure (traceless) version** This is more close to our subject : multipoles. 
$$
Y_\ell(\hat{n}) \propto r^{\ell + 1} (\hat{\mathbf{u}_1} \cdot \nabla) (\hat{\mathbf{u}_2} \cdot \nabla) \cdots (\hat{\mathbf{u}_\ell} \cdot \nabla)
$$

Writing each axis in spherical angles : 
$$
\hat{u}_k = (\sin{\theta_k} \cos{\varphi_k}, \sin{\theta_k} \sin{\varphi_k}, \cos{\theta_k})
$$
with $\theta_k \in [0, \pi]$ and $\varphi_k \in [0, 2\pi]$.
