# Mimicking polar decomposition in 3 dimensions

Despite the issues shown in *1_3d_issues.ipynb*, we can still try to decompose each order $\ell$ with a "polar-like" decomposition. However, this does not go very well... Let's break down why in this notebook. And also.. is that a big deal? 

## Maxwell axes

Represent a degree $\ell$-harmonic by $\ell$ axes $\hat{\mathbf{u}}, \dots, \mathbf{\hat{u}}_\ell$ plus a magnitude $r$.

**Raw (reducible) version.** The plain product of directional responses is then: 
$$g_{raw}(\hat{n}) = r \cdot \prod_{k=1}^{\ell}(\hat{n} \cdot \hat{u}_k)$$
That is a homogeneous degree-$\ell$ polynomial, so on the sphere it is a sum of harmonics of degrees $\ell, \ell-2, \ell-4, \dots$ (same parity) ; not pure degree $\ell$. 

However... let's stop here. Decompositions based on Maxwell angles are close from multipoles ; we'll then explore them comprehensively in *5_multipole.ipynb*.

## Hyperspherical coordinates

Let's see what happens by reformulating the system $\mathbf{a}_\ell \in \mathbb{R}^{2\ell + 1}$ with hyperspherical coordinates. Hence, we define the system

$$
\begin{cases}
a^1 & = r \cos{\varphi_1} \\
a^2 & = r \sin{\varphi_1} \cos{\varphi_2} \\
a^3 & = r \sin{\varphi_1} \sin{\varphi_2} \cos{\varphi_3} \\
& \vdots \\
a^{2\ell+1} & = r \sin{\varphi_1} \sin{\varphi_2} ... \sin{\varphi_{2\ell + 1}}
\end{cases}
$$

where $r \geq 0$, $\varphi_1, \dots, \varphi_{2\ell - 1} \in [0, \pi]$, and $\varphi_{2\ell} \in [0, 2\pi]$. This system can be inverted as follows : 

$$
\begin{cases}
r &= \Vert \mathbf{a}^\ell \Vert \\
\varphi_k &= \mathrm{atan}_2(\Vert a^{k+1}, \dots, a^{2_\ell+1} \Vert, a^{k}) \\
\varphi_{2\ell} &= \mathrm{atan}_2( a^{2\ell+1}, a^{2\ell}) \\
\end{cases}
$$

This works with $\ell=1$... but not above. Indeed, beyond this case where the charts are ordinary spherical coordinates of a 3-vector, the two angles directly match the source direction, and the coordinate sphere coincides with $S^2$. 

**Rotation behaviour**. Under a rotation, $a_\ell \mapsto D^\ell(R) a_\ell$. This preserves $r = ‖\mathbf{a}_\ell \Vert$, but for $\ell \geq 2$ it drags the point along a ≤ 3-dimensional orbit that is not aligned with the nested-sphere coordinate directions. Result: all $2\ell$ angles change in a coupled, nonlinear way; no single angle is "the orientation." Only $r$ is meaningful (invariant).

**Redundancy.** The redundancy here is the usual chart singularity of nested spheres : wherever $\sin \varphi_k = 0$ (a nested "pole"), the later angles $\varphi_{k+1}$ become undefined - the same gimbal-like collapse as $\theta=0$ in ordinary spherical coordinates.
