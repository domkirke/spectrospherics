# Mimicking polar decomposition in 3 dimensions

Despite the issues shown in *1_3d_issues.ipynb*, we can still try to decompose each order $\ell$ with a "polar-like" decomposition. However, this does not go very well... Let's break down why in this notebook. And also.. is that a big deal? 

## Maxwell axes

Represent a degree $\ell$-harmonic by $\ell$ axes $\hat{\mathbf{u}}, \dots, \mathbf{\hat{u}}_\ell$ plus a magnitude $r$.

**Raw (reducible) version.** The plain product of directional responses is then: 
$$g_{raw}(\hat{n}) = r \cdot \prod_{k=1}^{\ell}(\hat{n} \cdot \hat{u}_k)$$
That is a homogeneous degree-$\ell$ polynomial, so on the sphere it is a sum of harmonics of degrees $\ell, \ell-2, \ell-4, \dots$ (same parity) ; not pure degree $\ell$. 

However... let's stop here. Decompositions based on Maxwell angles are close from multipoles ; we'll then explore them comprehensively in *5_multipole.ipynb*. 

## Order-wise polar coordinates

The most straightforward solution is indeed, for a given degree $\ell$, to make $(|c_n^m|, \varphi_n^m$ couples) with each $(a^{-m}, a^{m})$ for $m \in [0, \ell]$. Hence, the $\ell$-decomposition of a spherical function is reformulated as follows

$$
\begin{align}
g(\theta, \varphi) = \sum_{n=0}^\ell \sum_{m=-n}^{n} a^m_n Y^m_n(\theta, \varphi)\\
\end{align}
$$

Within a given degree $n$ and an order pair ($-m, m$), we got : 
$$
\begin{align}
a^{-m}Y^{-m}_n + a^mY^m_n &= a^{-m}P^{|m|}_n(\cos \theta)  \sin {(|m| \phi)} + a^m P^{|m|}_n(\cos \theta)  \cos {(m \phi)} \\

 &= P^{m}_n(\cos \theta)  (a^{-m} \sin {(m \phi)} + a^m \cos {(m \phi)} ) \\
 &= P^{m}_n(\cos \theta) c^m e^{-jm\varphi} 
 
\end{align}
$$

where $|c^m| = \frac{1}{2}\sqrt{(a^{-m})^2+(a^{m})^2}$, $\phi^m = \mathrm{atan}_2(a^{-m}, a^m)$, and $c^m = |c^m|e^{-j\phi^m}$.  Hence we can re-write the decomposition as

$$
\begin{align}
g(\theta, \varphi) = a_n^0 P^0_n(\cos \theta) + \sum_{n=0}^\ell \sum_{m=0}^{n} P^{m}_\ell(\cos \theta) c^m_n e^{-jm\varphi} \\
\end{align}
$$

```python
from spectrospherics.spectrospherics import plot_pairwise_polar_spectrambisonics
# Play in hyperspherical coordinates! 
plot_pairwise_polar_spectrambisonics().show()
# plot_pairwise_polar_spectrambisonics().show()   # <- in browser version
```

This recombination has indeed several advantages :

- **It diagonalizes azimuthal rotation exactly at every degree.** Indeed, inntroducing a yaw shift $\alpha$ can be done by mapping phases $\phi^m_n \mapsto \phi^m_n + m \alpha. Hence, $r^m_n$ is indeed yaw invariant and $\phi^m_n$ is pure yaw-phase mode by mode. 
- **It is invertible without permutation ambiguity.** Unlike Maxwell-axis chart, ($|c^m_n|, $\phi^m_n$) is a smooth bijection (except usual singularity at $|c^m_n|=0$).

However, we can see that when more than 1 hamronic is enabled, shifting the phases $\phi^1_2$ and $(\phi^1_3)$ does not have a "rotation" role anymore, but can act on the shape or even in the tilt / roll of the surface. The problem is when other rotations besides yaw comes in to play. Unfortunately, for other rotations, $|c^m_n|$ is not invariant anymore : under a tilt or roll, the Wigner d-matric $D^\ell(R)$ mixes different $m$ within the degree over degree $\ell > 1$. 

It can be seen as a $SO(2)$ adpatation to 3D, but losing the "rotational" aspect of phase coefficients for $\ell>1$ if several harmonics are enabled.

**Why?** Well, we have to take back the small Wigner-D matrix formulation (see *B_wigner_d.ipynb*). Rememberng that the coefficients of a rotated sound field are linked by 

$$(a^m_n)' = \sum_{m=-n}^n D^n_{m'm}(R)a^m_n $$

Using the small formulation of Wigner d-matrix : 
$$
D^j_{m'm}(\alpha, \beta, \gamma) = e^{-im'\alpha}d^j_{m'm}(\beta)e^{-im\gamma}
$$
we can see that this mixing matrix has different shapes with different rotation angles = 
- For yaw, the rotation indeed comes down to a phase-shift similar to 2D : $D^j_{m'm}(\alpha, 0, 0) = e^{-im'\alpha}$
- For tilt / roll, the matrix is non-trivial anymore is not expressible as a phase component.

**It implies that it's not possible to express clearly these rotations with a phase.**. Hence, the "phase" aspect of coefficients for $\ell>1$ does not correspond to phase if several harmonics are enabled.

## Hyperspherical coordinates

Let's see what happens by reformulating the system $\mathbf{a}_\ell \in \mathbb{R}^{2\ell + 1}$ with hyperspherical coordinates. Hence, we define the system

$$
\begin{cases}
a^1 & = r \cos{\varphi_1} \\
a^2 & = r \sin{\varphi_1} \cos{\varphi_2} \\
a^3 & = r \sin{\varphi_1} \sin{\varphi_2} \cos{\varphi_3} \\
& \vdots \\
a^{2\ell} & = r \sin{\varphi_1} \sin{\varphi_2} ... \cos{\varphi_{2\ell}} \\
a^{2\ell+1} & = r \sin{\varphi_1} \sin{\varphi_2} ... \sin{\varphi_{2\ell}}
\end{cases}
$$

where $r \geq 0$, $\varphi_1, \dots, \varphi_{2\ell - 1} \in [0, \pi]$, and $\varphi_{2\ell} \in [0, 2\pi]$. This system can be inverted as follows : 

$$
\begin{cases}
r &= \Vert \mathbf{a}^\ell \Vert \\
\varphi_k &= \mathrm{atan}_2(\Vert a^{k+1}, \dots, a^{2_\ell+1} \Vert, a^{k}) \\
\varphi_{2\ell} &= \mathrm{atan}_2( a^{2\ell+1}, a^{2\ell}) \\
\end{cases}
$$

**Some faders do not do nothing? It's normal!** As the sinus of angles are ordered, if one is equal to $0$ it cancels the effects of all the angles that comes after... This is explained below. 


```python
from spectrospherics.spectrospherics import plot_hyperspherics_spectrambisonics
# Play in hyperspherical coordinates! 
plot_hyperspherics_spectrambisonics().servable()
# plot_hyperspherics_spectrambisonics().show()   # <- in browser version
```

 If you played a little bit around with radiuses and coefficients, you may have noticed some strange behavior... That's normal! 

**Rotation behaviour**. Under a rotation, $a_\ell \mapsto D^\ell(R) a_\ell$. This preserves $r = ‖\mathbf{a}_\ell \Vert$, but for $\ell \geq 2$ it drags the point along a ≤ 3-dimensional orbit that is not aligned with the nested-sphere coordinate directions. Result: all $2\ell$ angles change in a coupled, nonlinear way; no single angle is "the orientation" ; only $r$ is meaningful (invariant).See the graph below for an interactive representation of what's wrong for order 1.

**Gimbal collaphse.** The redundancy here is the usual chart singularity of nested spheres : wherever $\sin \varphi_k = 0$ (a nested "pole"), the later angles $\varphi_{k+1}$ become undefined - the same gimbal-like collapse as $\theta=0$ in ordinary spherical coordinates. This can be avoidable by not setting 0. but a tiny value, but this may still cause numerical problem when multiplying a lot of tiny values.
