# Wigner d-matrix


The link between rotations and spherical harmonics is explained by a single mathematical object : the **Wigner d-matrix**. This matrix is an unitary matrix in an irreducible representation of $SO(3)$ (or $SU(2)$; see *6_quaternions.ipynb*).

## Wigner D-matrix from Lie algebra. 

This is quite complex ; for a more extensive description of Lie algbera of $SO(3)$, please go to *B_so3_lie.ipynb*. 

Let $J_x$, $J_y$, $J_z$ be the generators of the Lie algbera of $SO(3)$ (also called *angular momentums$*). A 3-dimensional rotation operator can be written as 

$$
R(\alpha, \beta, \gamma) = e^{-i\alpha J_z}e^{-i \beta J_y} e^{-i\gamma J_z}
$$
where $(\alpha, \beta, \gamma)$ are the Euler angles in ZYZ convention. The Wigner D-Matrix is a unitary square matrix of dimensions $2j+1$ in this spherical basis with elements

$$
\begin{align}
D^j_{m'm} (\alpha, \beta, \gamma) & = \left \langle jm' | \mathcal{R}(\alpha, \beta, \gamma) | jm \right \rangle \\
& = e^{-im'\alpha} d^j_{m'm}(\beta) e^{-im\gamma}
\end{align}
$$

where 

$$\begin{align}
d^j_{m'm}(\beta) & = \left\langle jm' | e^{-i\beta J_y} | jm \right\rangle \\ & = D^j_{m'm}(0, \beta, 0)
\end{align}$$

is an element of the orthogonal Wigner's (small) d-matrix (sometimes referred to as the reduced Wigner D-matrix).

**OK, this is rough**. Let us see where that comes from.

## Orbits and stabilizers

Let's take the family $\mathcal{H}^\ell$ the family of degree $\ell$ harmonic polynomials $\approx$ the real spherical harmonics. We know that 

$$
\dim \mathcal{H}_\ell = 2 \ell + 1
$$

It can be shown that each spherical harmonic can be transformed in another one of same order by : 

$$
Y^m_l(\theta, \phi) = \sum_{m'=-l}^l [D^{(l)}_{mm'}(\mathcal{R})]^* Y^{m'}_l(\theta, \phi)
$$

where $D^{(l)}_{mm'}(\mathcal{R})$ is the Wigner D-matrix corresponding to the rotation defined on Euler angles $\mathcal{R}(\alpha, \beta, \gamma)$. 

Our rotation group, $SO(3)$, has 3 dimensions and acts on $\mathcal{H}_\ell$, the degree-$\ell$ spherical-harmonic coefficients $\mathbf{a}_\ell \in \mathbb{R}^{2\ell+1}$.

- An *orbit* (defined on a point) is the set you can reach from a point by acting with this group
$$ \mathcal{O}_\mathbf{a} = \{ R \mathbf{a} : R \in SO(3) \}$$

- A *stabilizer* (defined on a point) is the set of rotations leaving the point fixed
    

$$ \mathrm{Stab}(\mathbf{a}) = \{ R \in G : R \cdot \mathbf{a} = \mathbf{a} \} $$


- A *generic stabilizer* is a stabilizer shared by almost all points. For spherical harmonics : 

    - $\ell = 1$ : every point is axially symmetric, so the general stabilizer is $SO(2)$
    - $\ell \geq 2$ : a typical harmonic has no continuous symmetry — generic stabilizer is finite (dimension 0). Only special, highly-symmetric harmonics (zonal, cubic, icosahedral…) have bigger stabilizers, and they form thin exceptional strata. "Generic" means we ignore those.

For example, a zonal harmonic $Y^0_\ell$ is unchanged by any spin about $z$, such that its stabilizer constains $SO(2)$.

**Orbit dimension.** We can use orbits & stabilizers to quantify the quantities that do not change under the action. For that, we can retrieve the orbit dimension from the dimensions of the stabilizers : 
We then have : 
$$
\begin{align}
\mathcal{O}_\mathbf{a} & \approx G \backslash \mathrm{Stab}(\mathbf{a}) \\
\dim \mathcal{O}_\mathbf{a} &= \dim G - \dim \mathrm{Stab}(\mathbf{a})
\end{align}
$$

such that

$$
\dim \mathcal{O}_{\mathbf{a}} = 3 - \dim \mathrm{Stab}(\mathbf{a}) = \begin{cases} 0, \ & \ell = 0 \ (\text{fixed point}) \\2, \ & \ell = 1 \ (\text{Stab = }SO(2)\text{, orbit = }S^2) \\ 3, \ & \ell \geq 2 \ (\text{finite Stab, full-dim orbit)} \end{cases}
$$

where $\dim \mathcal{O}_{\mathbf{a}}$ can be seen as the cound of orientation degrees of freedom the field actually has. 

**Invariants** Hence, we can count the number of rotations that leave the field invariant. A function $f: V \rightarrow \mathbb{R}$ is invariant such that if $f(R\cdot \mathbf{a}) = f(\mathbf{a})$ for all $R$. Invariants are constant on each orbit, such that they coodrinatize the orbit space $\mathcal{H}_\ell \backslash G$ (the set of orbits). The number of independent invariants is the codimension of a generic orbit: 
$$\text{n\_invariants} = (2\ell + 1) - \dim \mathcal O$$
such that 
$$
\text{n\_invariants} = \begin{cases} 1, & \ell = 0, 1 \\ 2\ell - 2, & \ell \geq 2 \end{cases}
$$

Finally, we can summarize all of this with : 
$$
\underbrace{\dim \mathcal{H}_\ell}_{2\ell + 1 \text{ coefficients }} = \underbrace{\text{n\_invariants}}_{\text{untouched by rotation (shape / energy)}} + \underbrace{\dim \mathcal{O}}_{\text{reachable by rotation (orbit / orientation)}}
$$

**Consequences on coefficient redundency.** 

- **orbit dimensions** depending on degree $\ell$, the orbit dimension is either 0, 2, or 3. A rotation is defined with 3 numbers, sich that acting on a coefficient vector $\mathbf{a}_\ell \in \mathbb{R}^{2\ell+1}$ moves them into a $(2\ell + 1)$-dimensional space. 
- **stabilizer** : stabilizer $\mathrm{Stab}(\mathbf{a}) = \{R: D^\ell(R)\mathbf{a} = \mathbf{a} \}$ defines the *self-coincidence* : rotations that map the field onto itself.
