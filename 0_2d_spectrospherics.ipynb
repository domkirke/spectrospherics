# Spectrangular decomposition in 2d

Circular harmonics is defined by the series of orthogonal functions : 

$$
Y_m(\theta) = \begin{cases} \sin ( |m| \theta ) \ m \lt 0 \\ \cos(m\theta) \ m \geq 0 \end{cases}
$$

This series can be used to approiximate a sound field $g§\theta)$ through the projection onto this base with harmonic coefficients $(a_0, a_{-1}, a_1, ..., a_{-N}, a_N)$ : 

$$
\begin{align}
g(\theta)&= a_0Y_0(\theta) + \sum_{m=1}^N a_{-m}Y_{-m}(\theta) + a_m Y_{m}(\theta)\\
 & = a_0Y_0(\theta) + \sum_{m=1}^N a_{-m}\sin(|m| \theta) + a_m \cos (m\theta) \\
& = a_0Y_0(\theta) + \sum_{m=1}^N a_{-m} \frac{e^{im\theta} - e^{-im\theta}}{2i} + a_m \frac{e^{im\theta} + e^{-im\theta}}{2} \\
& = a_0Y_0(\theta) + \sum_{m=1}^N c_m e^{-im\theta} 
\end{align}
$$

with 
$$
\begin{cases}
c_m &= \frac{a_m + i a_{-m}}{2} \\
c_{-m} &= \frac{a_m - i a_{-m}}{2} = \bar{c_m}
\end{cases}
$$

that is a complex number with real part $a_m / 2$ and imaginary part $a_{-m}/2$. In polar form : 
$$
\begin{align}
|c_m| &= \sqrt{ \Big( \frac{a_m}{2} \Big)^2 + \Big( \frac{a_{-m}}{2} \Big)^2 } = \frac{1}{2} \sqrt{a^2_m + a^2_{-m}}\\
\phi_m &= \mathrm{arctan}_2(\frac{a_{-m}}{2}, \frac{a_m}{2}) = \mathrm{arctan}_2(a_{-m}, a_m)
\end{align}
$$

such that : 
$$c_m = |c_m| e^{i\phi_m}$$

Hence, we can have

$$
\begin{align}
g(\theta) &= c_o + \sum^N_{m=1} (c_m e^{-im\theta} + c_{-m} e^{-i (-m) \theta}) \\
& = \sum_{k=-N}^{k=N} c_k e^{-ik\theta}
\end{align}
$$

We then obtain a *spectrangular* formulation of $g(\theta)$, similar to a Fourier transform in the time domain by convolution of a kernel $e^{-im\theta}$. We can recover the original weights of circular harmonics.

$$
\begin{cases} a_m = c_m \cos{(2 \pi \phi_m)} \\ b_m = c_m \sin{(2 \pi \phi_m)} \end{cases}
$$

In the 2d case, the rotation operator $R(\alpha)$ acts as a pure phase component : 

$$ 
\begin{align}
c_k e^{-ik\theta} e^{-i\alpha} &= |c_m| e^{i\phi_m} e^{-ik\theta} e^{-i\alpha} \\
&= |c_m| e^{-i(\phi_m+\alpha)} e^{-ik\theta}
\end{align}
$$

such that $|a_m|$ is rotation-invariant, and $\phi_m$ is orientation only. We can also see that rotation group $SO(2)$ being abelian, implying that rotating by $\alpha$, and then by $\beta$, is commutative : 

$$
g(\theta)e^{-i\alpha}e^{-i\beta} = g(\theta)e^{-i\beta}e^{-i\alpha}
$$

Using notations $R(\alpha)$ and $R(\beta)$ for, respectively, a rotation of angle $\alpha$ and $\beta$, commutativty can the be writen :

$$
R(\alpha)R(\beta) = R(\alpha + \beta) = R(\beta)R(\alpha)
$$
